```python
# Goal : Predict the Calories Burned During a Workout session
```

```python
df.columns
```

```python
cols = ["Age" , "Gender" , "Weight (kg)" , "Height (m)" , "Session_Duration (hours)", 'Workout_Frequency (days/week)', 'Experience_Level'  ,"Calories_Burned"]
prediction_df = df[cols]
```

```python
prediction_df
```

```python
prediction_df.columns = [
    "age" ,
    "gender" ,
    "weight" ,
    "height" ,
    "duration" ,
    "frequency",
    "experience",
    "calories"
]
prediction_df
```

```python
def preprocess(df):

  new_df = df.copy()

  new_df["gender"] = new_df["gender"].apply(lambda x : 1 if x == "Male" else 0)


  return new_df


```

```python
#split the dataset
from sklearn.model_selection import train_test_split

X_train , X_test , y_train , y_test = train_test_split(
    prediction_df.drop(columns=["calories"]),
    prediction_df["calories"],
    test_size=0.2,
    random_state=42
)
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
X_train_scaled = scaler.fit_transform(preprocess(X_train))
X_test_scaled = scaler.transform(preprocess(X_test))
```

```python
#fit linear regression
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(X_train_scaled , y_train)
```

```python
from sklearn.metrics import mean_squared_error , mean_absolute_error , r2_score, root_mean_squared_error
def evaluate(model, X_test , y_test):
  y_pred = model.predict(X_test)
  return {
      "R2 Score" : r2_score(y_test , y_pred),
      "MAE" : mean_absolute_error(y_test , y_pred),
      "MSE" : mean_squared_error(y_test , y_pred),
      "RMSE" : root_mean_squared_error(y_test , y_pred)
  }
```

```python
evaluate(model, X_test_scaled , y_test)
```

```python

def predict_calories(age, gender, weight, height, duration):

  df_predict = pd.DataFrame({
      "age" : [age],
      "gender" : [gender],
      "weight" : [weight],
      "height" : [height],
      "duration" : [duration]
  })

  df_predict = preprocess(df_predict)
  df_predict_scaled = scaler.transform(df_predict)

  return model.predict(df_predict_scaled)[0]

```

```python
predict_calories(
    age=30,
    gender="Male",
    weight=70,
    height=1.75,
    duration=2
)
```
