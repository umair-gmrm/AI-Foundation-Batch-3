# Functions

```python
def area_of_circle(radius):
  PI = 3.14
  area =  PI * (radius ** 2)
  print(f"Area of a circle with the radius of {radius} is {area}")
  return area
```

```python
area_of_circle(10)
area_of_circle(50)
```

```python
def area_of_square(length):
  area = (length ** 2)
  return area
```

```python
area_of_circle(1) + area_of_circle(2)
```

```python
area_of_square(10) + area_of_square(5)
```

```python
def area_of_rect(length , width):
  return length * width
```

```python
area_of_rect(10, 20)
```
